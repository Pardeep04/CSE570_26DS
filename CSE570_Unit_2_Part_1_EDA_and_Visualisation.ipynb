## Load the dataset
This cell works locally and in Google Colab.

```python
# This cell makes the notebook work both locally and in Google Colab.
from pathlib import Path
import pandas as pd

csv_name = "student_performance_eda.csv"
csv_path = Path(csv_name)

if not csv_path.exists():
    try:
        from google.colab import files
        print(f"Please upload {csv_name}")
        uploaded = files.upload()
        csv_path = Path(next(iter(uploaded.keys())))
    except ImportError:
        raise FileNotFoundError(
            f"{csv_name} was not found. Place it in the same folder as this notebook."
        )

df = pd.read_csv(csv_path)
print("Dataset loaded successfully.")
print("Shape:", df.shape)
df.head()
```

## Basic inspection

```python
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
df.info()
```

```python
df.describe(include='all').round(2)
```

## Missing values and duplicates

```python
missing_report = pd.DataFrame({
    "Missing_Count": df.isnull().sum(),
    "Missing_Percentage": df.isnull().mean().mul(100).round(2)
})
missing_report
```

```python
print("Duplicate rows:", df.duplicated().sum())
print("Repeated IDs:", df["Student_ID"].duplicated().sum())
```

## Categorical distributions

```python
for col in ["Gender", "City", "Study_Mode", "Internet_Access", "Placed"]:
    print("\n", col)
    print(df[col].value_counts(dropna=False))
```

## Histogram

```python
plt.figure(figsize=(8,5))
plt.hist(df["Final_Score"].dropna(), bins=12, edgecolor="black")
plt.title("Distribution of Final Scores")
plt.xlabel("Final Score")
plt.ylabel("Number of Students")
plt.show()
```

## Bar chart

```python
counts = df["Study_Mode"].value_counts()
plt.figure(figsize=(7,4))
plt.bar(counts.index, counts.values)
plt.title("Students by Study Mode")
plt.xlabel("Study Mode")
plt.ylabel("Count")
plt.show()
```

## Box plot

```python
plt.figure(figsize=(7,4))
plt.boxplot(df["Study_Hours_Per_Day"].dropna(), vert=False)
plt.title("Study Hours per Day")
plt.xlabel("Hours")
plt.show()
```

## Scatter plot

```python
plot_data = df.dropna(subset=["Study_Hours_Per_Day", "Final_Score"])
plt.figure(figsize=(8,5))
plt.scatter(plot_data["Study_Hours_Per_Day"], plot_data["Final_Score"], alpha=0.7)
plt.title("Study Hours and Final Score")
plt.xlabel("Study Hours per Day")
plt.ylabel("Final Score")
plt.show()
```
