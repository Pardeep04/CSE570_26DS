# Random Forest Classification

## Importing the libraries

```
# Pandas is used to read and work with the CSV dataset.
import pandas as pd
```

## Importing the dataset

```
# Read the social network advertisements dataset from a CSV file.
dataset = pd.read_csv('Social_Network_Ads.csv')

# Select Age and Estimated Salary (columns 2 and 3) as input features.
X = dataset.iloc[:, [2, 3]].values

# Select the final column (Purchased) as the target class.
y = dataset.iloc[:, -1].values
```

## Splitting the dataset into the Training set and Test set

```
from sklearn.model_selection import train_test_split

# Use 75% of the data for training and 25% for testing.
# random_state=0 makes the split reproducible.
X_train, X_test, y_train, y_test = train_test_split(
    X, y, test_size=0.25, random_state=0
)
```

## Feature Scaling

```
from sklearn.preprocessing import StandardScaler

# Standardize features using the mean and standard deviation of the training data.
# Fit only on training data to avoid leaking test-set information.
sc = StandardScaler()
X_train = sc.fit_transform(X_train)
X_test = sc.transform(X_test)
```

## Training the Random Forest Classification model on the Training set

```
from sklearn.ensemble import RandomForestClassifier

# Build a Random Forest containing 10 decision trees.
# 'entropy' measures information gain when deciding how to split nodes.
# random_state=0 makes the model results reproducible.
classifier = RandomForestClassifier(
    n_estimators=10, criterion='entropy', random_state=0
)

# Train the model using the scaled training features and class labels.
classifier.fit(X_train, y_train)
```

## Predicting the Test set results

```
# Predict whether each user in the test set belongs to the Purchased class.
y_pred = classifier.predict(X_test)
```

## Making the Confusion Matrix

```
from sklearn.metrics import confusion_matrix

# Compare actual labels with predicted labels.
# For binary labels 0 and 1, rows are actual classes and columns are predictions.
# The diagonal contains correct predictions; off-diagonal values are errors.
cm = confusion_matrix(y_test, y_pred)
print(cm)
```
