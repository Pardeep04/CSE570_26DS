```python
# Import NumPy
import numpy as np

print("NumPy version:", np.__version__)
```

```python
# Variables and common data types
student_name = "Aman"
age = 21
percentage = 82.5
is_eligible = True

print(student_name, type(student_name))
print(age, type(age))
print(percentage, type(percentage))
print(is_eligible, type(is_eligible))
```

```python
# Type conversion
age_text = "22"
age_number = int(age_text)

price = 199
price_decimal = float(price)

print(age_number, type(age_number))
print(price_decimal, type(price_decimal))
```

```python
# String values must be converted before numerical addition
first_value = "20"
second_value = "5"

total = int(first_value) + int(second_value)
print(total)
```

```python
# Arithmetic operators
a = 15
b = 4

print("Addition:", a + b)
print("Subtraction:", a - b)
print("Multiplication:", a * b)
print("Division:", a / b)
print("Floor division:", a // b)
print("Remainder:", a % b)
print("Power:", a ** b)
```

```python
# Comparison and logical operators
marks = 75
attendance = 82

passed_exam = marks >= 40
has_attendance = attendance >= 75
is_eligible = passed_exam and has_attendance

print("Passed:", passed_exam)
print("Attendance complete:", has_attendance)
print("Eligible:", is_eligible)
```

```python
# Conditional statement
marks = 68

if marks >= 75:
    grade = "Distinction"
elif marks >= 60:
    grade = "First Division"
elif marks >= 40:
    grade = "Pass"
else:
    grade = "Fail"

print("Grade:", grade)
```

```python
# Lists
marks = [72, 85, 68, 90, 77]

print("First:", marks[0])
print("Last:", marks[-1])
print("First three:", marks[0:3])

marks.append(88)
marks[2] = 70

print("Updated list:", marks)
```

```python
# Useful list functions
marks = [72, 85, 68, 90, 77]

print("Count:", len(marks))
print("Maximum:", max(marks))
print("Minimum:", min(marks))
print("Total:", sum(marks))
print("Average:", sum(marks) / len(marks))
```

```python
# Tuple: values cannot be changed after creation
coordinates = (30.90, 75.85)

print("Latitude:", coordinates[0])
print("Longitude:", coordinates[1])
```

```python
# Dictionary
student = {
    "id": 101,
    "name": "Aman",
    "marks": 82
}

print(student["name"])

student["city"] = "Jalandhar"
student["marks"] = 86

print(student)
```

```python
# for loop
marks = [72, 85, 68, 90]

for mark in marks:
    print("Current mark:", mark)
```

```python
# Loop with condition
marks = [72, 35, 68, 29, 90]

for mark in marks:
    if mark >= 40:
        print(mark, "-> Pass")
    else:
        print(mark, "-> Fail")
```

```python
# while loop
count = 1

while count <= 5:
    print("Count:", count)
    count += 1
```

```python
# Function
def calculate_average(values):
    return sum(values) / len(values)

student_marks = [72, 85, 68, 90]
result = calculate_average(student_marks)

print("Average:", result)
```

```python
# Create a 1-D NumPy array
marks = np.array([72, 85, 68, 90, 77])

print(marks)
print("dtype:", marks.dtype)
print("ndim:", marks.ndim)
print("shape:", marks.shape)
print("size:", marks.size)
```

```python
# Create a 2-D NumPy array
student_data = np.array([
    [101, 72, 80],
    [102, 85, 88],
    [103, 68, 75]
])

print(student_data)
print("Dimensions:", student_data.ndim)
print("Shape:", student_data.shape)
```

```python
# Special NumPy arrays
zeros_array = np.zeros(5)
ones_array = np.ones((2, 3))
range_array = np.arange(0, 10, 2)
linear_array = np.linspace(0, 1, 5)

print("Zeros:", zeros_array)
print("Ones:\n", ones_array)
print("Range:", range_array)
print("Equally spaced:", linear_array)
```

```python
# Indexing and slicing
values = np.array([10, 20, 30, 40, 50])

print(values[0])
print(values[-1])
print(values[1:4])
print(values[:3])
print(values[::2])
```

```python
# 2-D indexing
matrix = np.array([
    [10, 20, 30],
    [40, 50, 60],
    [70, 80, 90]
])

print("Row 2, column 3:", matrix[1, 2])
print("First row:", matrix[0, :])
print("Second column:", matrix[:, 1])
```

```python
# Vectorised operations
marks = np.array([72, 85, 68, 90])

bonus_marks = marks + 5
proportions = marks / 100
squared_marks = marks ** 2

print("Bonus:", bonus_marks)
print("Proportions:", proportions)
print("Squares:", squared_marks)
```

```python
# Aggregate functions
marks = np.array([72, 85, 68, 90, 77])

print("Sum:", np.sum(marks))
print("Mean:", np.mean(marks))
print("Median:", np.median(marks))
print("Minimum:", np.min(marks))
print("Maximum:", np.max(marks))
print("Standard deviation:", np.std(marks))
```

```python
# Reshaping
values = np.array([1, 2, 3, 4, 5, 6])
reshaped = values.reshape(2, 3)

print(reshaped)
```

```python
# Boolean filtering
marks = np.array([72, 35, 68, 29, 90])

pass_mask = marks >= 40
passing_marks = marks[pass_mask]

print("Mask:", pass_mask)
print("Passing marks:", passing_marks)
```

```python
# Missing numerical values with NumPy
scores = np.array([72, 85, np.nan, 90, 77])

print("Regular mean:", np.mean(scores))
print("Mean ignoring NaN:", np.nanmean(scores))
print("Missing mask:", np.isnan(scores))
```

```python
# Mini practice program
marks = np.array([78, 92, 35, 64, 88])

average_mark = np.mean(marks)
highest_mark = np.max(marks)
lowest_mark = np.min(marks)
passing_marks = marks[marks >= 40]

print("All marks:", marks)
print("Average:", average_mark)
print("Highest:", highest_mark)
print("Lowest:", lowest_mark)
print("Passing marks:", passing_marks)
```
