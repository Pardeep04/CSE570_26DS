# Logistic Regression — Explained Notebook for Beginners

**Learning objective:** Use a person's **Age** and **Estimated Salary** to predict whether they **Purchased** a product (`1`) or **Did Not Purchase** (`0`). This is **binary classification**, not salary prediction.

**Run the notebook in order from top to bottom.** Keep `Social_Network_Ads.csv` in the same working folder as the notebook, or adjust the file path in the data-loading cell.

> **Color guide:** 🔵 concept • 🟢 example • 🟠 caution • 🟣 formula

🟣 **Core formula:** `z = b0 + b1*x1 + b2*x2`, then `p = 1 / (1 + e^(−z))`. The sigmoid returns a value between 0 and 1, interpreted as the model's estimated probability of class 1. With a 0.5 threshold, `p >= 0.5` produces class 1; otherwise class 0.


# Logistic Regression

## Importing the libraries

> 🔵 **Student note — before running this cell**
>
> **What this cell does:** Loads three useful Python libraries. `numpy` handles numerical arrays, `matplotlib.pyplot` draws graphs, and `pandas` reads table-like data. `as np`, `as plt`, and `as pd` are short names used later.


```
# NumPy: arrays and numerical calculations
import numpy as np
# Matplotlib: drawing charts and decision regions
import matplotlib.pyplot as plt
# Pandas: loading and working with tabular CSV data
import pandas as pd
```

## Importing the dataset

> 🔵 **Student note — before running this cell**
>
> **Understanding the dataset:** `pd.read_csv()` opens the CSV file. `iloc` selects rows and columns by *position* (counting from 0). `:` means all rows. Columns 2 and 3 become the input features `X` (Age and Estimated Salary in this dataset). The last column (`-1`) becomes the target `y` (whether a user purchased: 0 = No, 1 = Yes). `.values` converts selected data into NumPy arrays. The CSV must be located in the notebook's working directory, or you must provide its path.


```
# Read the advertising dataset from the current working folder
dataset = pd.read_csv('Social_Network_Ads.csv')
# Choose every row and columns at positions 2 and 3 as features (Age, Salary)
X = dataset.iloc[:, [2, 3]].values
# Choose the last column as the target class (Purchased: 0 or 1)
y = dataset.iloc[:, -1].values
```

## Splitting the dataset into the Training set and Test set

> 🔵 **Student note — before running this cell**
>
> **Why split the data?** The training set teaches the model; the test set checks performance on examples it has not seen during training. `test_size=0.25` assigns 25% of rows to testing and 75% to training. `random_state=0` makes the random split repeatable. `X` stores input columns, while `y` stores the correct answers.


```
# Import the function for separating examples into training and test groups
from sklearn.model_selection import train_test_split
# Keep 25% for testing; random_state makes the same split reproducible
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size = 0.25, random_state = 0)
```

## Feature Scaling

> 🔵 **Student note — before running this cell**
>
> **Why standardize?** Age and salary are measured on different numerical scales. `StandardScaler` transforms each feature approximately as `(value − training mean) / training standard deviation`. `fit_transform(X_train)` learns the training-set mean and standard deviation and applies them; `transform(X_test)` uses *the same* values on the test set. This avoids using information from test data during training.


```
# Import feature standardization (mean 0 and standard deviation ~1 on training)
from sklearn.preprocessing import StandardScaler
# Create the scaler object
sc = StandardScaler()
# Learn scaling values ONLY from training data, then scale training inputs
X_train = sc.fit_transform(X_train)
# Apply those SAME training scaling values to unseen test inputs
X_test = sc.transform(X_test)
```

## Training the Logistic Regression model on the Training set

> 🔵 **Student note — before running this cell**
>
> **How the model learns:** `LogisticRegression()` creates a binary classifier. `.fit(X_train, y_train)` learns feature coefficients and an intercept from the training data. It first combines inputs linearly, `z = b0 + b1 × age + b2 × salary` (using the scaled feature values), then the sigmoid function turns `z` into a probability: `p = 1 / (1 + exp(−z))`. For binary classification, the usual decision threshold is 0.5.


```
# Import the Logistic Regression classifier
from sklearn.linear_model import LogisticRegression
# Create the model; random_state is included for repeatability where applicable
classifier = LogisticRegression(random_state = 0)
# Fit the model: learn how Age and Salary relate to Purchased
classifier.fit(X_train, y_train)
```

## Predicting the Test set results

> 🔵 **Student note — before running this cell**
>
> **Prediction:** `.predict(X_test)` outputs class labels (0 or 1), *not* percentages. `y_pred` stores those predictions. To inspect probabilities instead, you can use `classifier.predict_proba(X_test)`; its columns correspond to `classifier.classes_`.


```
# Predict a 0 or 1 label for each unseen test example
y_pred = classifier.predict(X_test)
```

## Making the Confusion Matrix

> 🔵 **Student note — before running this cell**
>
> **Confusion matrix:** This compares real labels (`y_test`) against predicted labels (`y_pred`). For binary classes ordered `[0, 1]`, the matrix is `[[TN, FP], [FN, TP]]`. TN = correctly predicted No; FP = predicted Yes but actually No; FN = predicted No but actually Yes; TP = correctly predicted Yes. Correct predictions lie on the main diagonal. Accuracy can be computed as `(TN + TP) / total`.


```
# Import the tool for comparing true and predicted labels
from sklearn.metrics import confusion_matrix
# Rows = actual classes, columns = predicted classes
cm = confusion_matrix(y_test, y_pred)
# Display the confusion matrix: [[TN, FP], [FN, TP]] for classes [0, 1]
print(cm)
```
