```python
# Create a DataFrame
student_data = {
    "StudentID": [101, 102, 103, 104],
    "Name": ["Aman", "Riya", "Karan", "Simran"],
    "Age": [20, 21, 20, 22],
    "Marks": [78, 85, 69, 91]
}

df = pd.DataFrame(student_data)
print(df)
```

```python
# Select columns
print(df["Name"])
print(df[["Name", "Marks"]])
```

```python
# Select rows using loc and iloc
print(df.loc[0])
print(df.loc[0:2, ["Name", "Marks"]])

print(df.iloc[0])
print(df.iloc[0:3, 1:4])
```

```python
# Filter rows
high_scorers = df[df["Marks"] >= 80]
print(high_scorers)
```

```python
# Filter using multiple conditions
filtered_students = df[
    (df["Marks"] >= 75) &
    (df["Age"] <= 21)
]

print(filtered_students)
```

```python
# Add new columns
df["Passed"] = df["Marks"] >= 40
df["BonusMarks"] = df["Marks"] + 5

print(df)
```

```python
# Update a particular value
df.loc[df["StudentID"] == 103, "Marks"] = 74
print(df)
```

```python
# Remove a column
df_without_bonus = df.drop(columns=["BonusMarks"])
print(df_without_bonus)
```

```python
# Remove a row
df_without_first_row = df.drop(index=0)
print(df_without_first_row)
```

```python
# Sort values
sorted_df = df.sort_values(by="Marks", ascending=False)
print(sorted_df)
```

```python
# Sort using multiple columns
sorted_df = df.sort_values(
    by=["Age", "Marks"],
    ascending=[True, False]
)

print(sorted_df)
```

```python
# Rename a column
renamed_df = df.rename(columns={"Marks": "FinalMarks"})
print(renamed_df)
```

```python
# Create a sample dataset for inspection
data = {
    "StudentID": [101, 102, 103, 104, 105],
    "Age": [20, 21, np.nan, 22, 20],
    "City": ["Delhi", "Mumbai", "Delhi", None, "Chandigarh"],
    "Marks": [78, np.nan, 69, 91, 85]
}

df = pd.DataFrame(data)
print(df)
```

```python
# Preview the dataset
print(df.head())
print(df.head(10))
print(df.tail())
print(df.sample(min(5, len(df)), random_state=42))
```

```python
# Shape, columns and data types
print("Shape:", df.shape)
print("Columns:", df.columns.tolist())
print(df.dtypes)
```

```python
# Dataset information
df.info()
```

```python
# Descriptive statistics
print(df.describe())
print(df.describe(include="all"))
```

```python
# Unique values and frequency counts
print(df.nunique())
print(df["City"].unique())
print(df["City"].value_counts())
print(df["City"].value_counts(dropna=False))
```

```python
# Detect duplicate rows
print(df.duplicated())
print("Duplicate rows:", df.duplicated().sum())

df_no_duplicates = df.drop_duplicates()
print(df_no_duplicates)
```

```python
# Detect missing values
print(df.isna())
print(df.isna().sum())
```

```python
# Missing-value report
missing_percentage = (df.isna().sum() / len(df)) * 100

missing_report = pd.DataFrame({
    "Missing_Count": df.isna().sum(),
    "Missing_Percentage": missing_percentage.round(2)
})

print(missing_report)
```

```python
# Remove rows containing missing values
complete_rows = df.dropna()
print(complete_rows)
```

```python
# Remove rows only when Marks is missing
df_marks_available = df.dropna(subset=["Marks"])
print(df_marks_available)
```

```python
# Remove columns containing missing values
df_complete_columns = df.dropna(axis=1)
print(df_complete_columns)
```

```python
# Mean imputation
df_mean = df.copy()
marks_mean = df_mean["Marks"].mean()
df_mean["Marks"] = df_mean["Marks"].fillna(marks_mean)

print(df_mean)
```

```python
# Median imputation
df_median = df.copy()
age_median = df_median["Age"].median()
df_median["Age"] = df_median["Age"].fillna(age_median)

print(df_median)
```

```python
# Mode imputation for categorical data
df_mode = df.copy()
city_mode = df_mode["City"].mode()[0]
df_mode["City"] = df_mode["City"].fillna(city_mode)

print(df_mode)
```

```python
# Use an explicit category for missing categorical values
df_unknown = df.copy()
df_unknown["City"] = df_unknown["City"].fillna("Unknown")

print(df_unknown)
```

```python
# Forward fill and backward fill
df_forward = df.ffill()
df_backward = df.bfill()

print("Forward fill:\n", df_forward)
print("\nBackward fill:\n", df_backward)
```

```python
# Missing-value indicator
df_indicator = df.copy()

df_indicator["Marks_Was_Missing"] = (
    df_indicator["Marks"].isna().astype(int)
)

df_indicator["Marks"] = df_indicator["Marks"].fillna(
    df_indicator["Marks"].median()
)

print(df_indicator)
```

```python
# Imputation using Scikit-Learn
from sklearn.impute import SimpleImputer

df_imputed = df.copy()

numeric_imputer = SimpleImputer(strategy="median")
df_imputed[["Age", "Marks"]] = numeric_imputer.fit_transform(
    df_imputed[["Age", "Marks"]]
)

categorical_imputer = SimpleImputer(strategy="most_frequent")
df_imputed[["City"]] = categorical_imputer.fit_transform(
    df_imputed[["City"]]
)

print(df_imputed)
```

```python
# Complete cleaning example
raw_data = {
    "StudentID": [101, 102, 102, 103, 104],
    "Name": [" Aman ", "Riya", "Riya", "Karan", "Simran"],
    "Age": [20, 21, 21, np.nan, 22],
    "City": ["Delhi", "Mumbai", "Mumbai", None, "Delhi"],
    "Marks": [78, 85, 85, 69, np.nan]
}

clean_df = pd.DataFrame(raw_data)

clean_df = clean_df.drop_duplicates()
clean_df["Name"] = clean_df["Name"].str.strip()
clean_df["Age"] = clean_df["Age"].fillna(clean_df["Age"].median())
clean_df["Marks"] = clean_df["Marks"].fillna(clean_df["Marks"].median())
clean_df["City"] = clean_df["City"].fillna(clean_df["City"].mode()[0])
clean_df = clean_df.reset_index(drop=True)

print(clean_df)
```
